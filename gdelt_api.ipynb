```python
# Mentions endpoint
import pandas as pd

latest = pd.read_csv(
    "http://data.gdeltproject.org/gdeltv2/lastupdate.txt",
    sep=" ",
    header=None
)

mentions_url = latest.loc[
    latest[2].str.contains("mentions", na=False),
    2
].iloc[0]

print(mentions_url)

mentions = pd.read_csv(
    mentions_url,
    sep="\t",
    header=None,
    compression="zip",
    low_memory=False
)

print(mentions.head())
print(mentions.shape)
```

```python
# GKG endpoint
import pandas as pd

latest = pd.read_csv(
    "http://data.gdeltproject.org/gdeltv2/lastupdate.txt",
    sep=" ",
    header=None
)

gkg_url = latest.loc[
    latest[2].str.contains("gkg", na=False),
    2
].iloc[0]

print(gkg_url)

gkg = pd.read_csv(
    gkg_url,
    sep="\t",
    header=None,
    compression="zip",
    low_memory=False
)

print(gkg.head())
print(gkg.shape)
```
